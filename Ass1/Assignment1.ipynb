# Deep Learning Assignment 1
**Objective**: Install and configure TensorFlow/Keras, perform data preprocessing, normalization, train-test split, and visualization using the **Fashion MNIST** dataset.

```python
# 1. Import libraries and check TensorFlow version
import tensorflow as tf
import matplotlib.pyplot as plt
import numpy as np

print("TensorFlow Version:", tf.__version__)
```

## 2. Load the Dataset and Perform Train-Test Split
The Fashion MNIST dataset is built into Keras. The `load_data()` function automatically splits it into training and testing sets.

```python
fashion_mnist = tf.keras.datasets.fashion_mnist
(train_images, train_labels), (test_images, test_labels) = fashion_mnist.load_data()

print("Training images shape:", train_images.shape)
print("Training labels shape:", train_labels.shape)
print("Testing images shape:", test_images.shape)
print("Testing labels shape:", test_labels.shape)
```

## 3. Define Class Names
The labels are integers from 0 to 9. We map them to their actual class names.

```python
class_names = [
    'T-shirt/top', 'Trouser', 'Pullover', 'Dress', 'Coat',
    'Sandal', 'Shirt', 'Sneaker', 'Bag', 'Ankle boot'
]

print("First image label:", train_labels[0])
print("Class name:", class_names[train_labels[0]])
```

## 4. Data Visualization (Before Normalization)
Let's visualize the first image to see what the pixel values look like. Notice the colorbar ranges from 0 to 255.

```python
plt.figure(figsize=(4,4))
plt.imshow(train_images[0], cmap='gray')
plt.title(class_names[train_labels[0]])
plt.colorbar()
plt.show()
```

## 5. Data Normalization
Neural networks process values between 0 and 1 much better than 0 to 255. We normalize the image data by dividing by 255.0.

```python
train_images = train_images / 255.0
test_images = test_images / 255.0

print("Data normalized successfully!")
```

## 6. Verify the Data Format
Let's display the first 25 images from the training set and display the class name below each image to verify that the data is in the correct format.

```python
plt.figure(figsize=(10,10))
for i in range(25):
    plt.subplot(5,5,i+1)
    plt.xticks([])
    plt.yticks([])
    plt.grid(False)
    plt.imshow(train_images[i], cmap='gray')
    plt.title(class_names[train_labels[i]])
plt.tight_layout()
plt.show()
```
